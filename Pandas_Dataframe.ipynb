# Pandas Dataframe




Q1. Creating a DataFrame

```python
import pandas as pd

# Create a sample DataFrame using a Python dictionary
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Printer'],
    'Region': ['North', 'South', 'East', 'West', 'Central'],
    'Sales': [45000, 1500, 2500, 12000, 8000],
    'Quantity': [5, 20, 15, 8, 4]
}

df_q1 = pd.DataFrame(data)

# Display the DataFrame
print("--- Q1: Sample DataFrame ---")
print(df_q1)
```

Q2. Data Inspection

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Retail_dataset.csv")

# 1. Display the first 5 rows of the dataset
print("--- First 5 Rows ---")
print(df.head())

# 2. Display the last 5 rows
print("\n--- Last 5 Rows ---")
print(df.tail())

# 3. Show the structure of the dataset using info()
print("\n--- Dataset Info ---")
df.info()

# 4. Display the shape of the dataset
print("\n--- Shape (Rows, Columns) ---")
print(df.shape)
```

Q3. Column Selection

```python

# Display the column names first to check exact names
print("--- Available Columns ---")
print(df.columns)

# Select specific columns (replace these with actual column names from your dataset if needed)
# Example: selecting two or more columns
selected_columns = df[['Product', 'Sales']]

print("\n--- Selected Columns (First 5 Rows) ---")
print(selected_columns.head())
```

Q4. Indexing using .loc

```python
# 1. Display the row with index 10
print("--- Row with Index 10 ---")
print(df.loc[10])

# 2. Display the Sales value of the row with index 20
print("\n--- Sales Value at Index 20 ---")
print(df.loc[20, 'Sales'])
```

Q5. Indexing using .iloc

```python
# 1. Display the first 6 rows and first 4 columns
print("--- First 6 Rows and First 4 Columns ---")
print(df.iloc[:6, :4])

# 2. Display the Sales value from the 3rd row (index 2 in 0-based indexing)
print("\n--- Sales Value from the 3rd Row ---")
print(df.iloc[2]['Sales'])
```

Q6. Row Slicing

```python
# 1. Display the first 10 rows of the dataset
print("--- First 10 Rows ---")
print(df.iloc[:10])

# 2. Display rows 15 to 25
print("\n--- Rows 15 to 25 ---")
print(df.iloc[15:26])
```

Q7. Filtering Data

```python
# 1. Filter rows where Region is "East"
east_region_df = df[df['Region'] == 'East']
print("--- Rows where Region is 'East' ---")
print(east_region_df.head())

# 2. Filter rows where Sales is greater than 1500
high_sales_df = df[df['Sales'] > 1500]
print("\n--- Rows where Sales > 1500 ---")
print(high_sales_df.head())
```

Q8. Multiple Condition Filtering

```python
# Filter rows where Region is "West" and Sales is greater than 1200
multiple_filtered_df = df[(df['Region'] == 'West') & (df['Sales'] > 1200)]

# Display the resulting rows
print("--- Region is 'West' and Sales > 1200 ---")
print(multiple_filtered_df)
```

Q9. Descriptive Statistics

```python
# Generate summary statistics using describe()
stats = df.describe()
print("--- Summary Statistics ---")
print(stats)

# Extract specific metrics requested
print("\n--- Key Metrics ---")
print("Mean Sales Value:", df['Sales'].mean())
print("Maximum Quantity Sold:", df['Quantity'].max())
print("Minimum Sales Value:", df['Sales'].min())
```

Q10. Aggregation Analysis

```python
# Calculate aggregation metrics using the Sales column
total_sales = df['Sales'].sum()
average_sales = df['Sales'].mean()
maximum_sales = df['Sales'].max()
minimum_sales = df['Sales'].min()

# Display the aggregation results
print("--- Sales Aggregation Analysis ---")
print(f"1. Total Sales: {total_sales}")
print(f"2. Average Sales: {average_sales}")
print(f"3. Maximum Sales: {maximum_sales}")
print(f"4. Minimum Sales: {minimum_sales}")
```
